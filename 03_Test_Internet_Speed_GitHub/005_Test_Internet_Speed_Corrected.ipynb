## 2. Test Internet Speed Using Python

```python
import speedtest

st = speedtest.Speedtest()

download = st.download()
upload = st.upload()

download = download / 1_000_000
upload = upload / 1_000_000

st.get_servers([])
ping = st.results.ping

print("Your ⏬ Download speed is", round(download, 3), "Mbps")
print("Your ⏫ Upload speed is", round(upload, 3), "Mbps")
print("Your Ping is", round(ping, 2), "ms")
```

## 3. Internet Speed Test Using Tkinter GUI

The icon line is optional. Leave it commented unless `speed.ico` exists.

```python
import speedtest
import threading
import tkinter as tk
from tkinter import ttk, messagebox


class InternetSpeedTestGUI:
    def __init__(self, root):
        self.root = root
        self.root.title("Test Internet Speed")
        self.root.geometry("380x260")
        self.root.resizable(False, False)
        self.root.configure(bg="#ffffff")

        # Optional icon:
        # self.root.iconbitmap("speed.ico")

        tk.Label(
            root,
            text="TEST INTERNET SPEED",
            bg="#ffffff",
            fg="#404042",
            font=("Arial", 23, "bold")
        ).pack()

        tk.Label(
            root,
            text="Python Internet Speed Tester",
            bg="#ffffff",
            fg="#404042",
            font=("Arial", 10, "bold")
        ).pack(side=tk.BOTTOM, pady=10)

        self.down_label = tk.Label(
            root, text="⏬ Download Speed - ",
            bg="#ffffff", font=("Arial", 10, "bold")
        )
        self.down_label.place(x=90, y=50)

        self.up_label = tk.Label(
            root, text="⏫ Upload Speed - ",
            bg="#ffffff", font=("Arial", 10, "bold")
        )
        self.up_label.place(x=90, y=80)

        self.ping_label = tk.Label(
            root, text="Your Ping - ",
            bg="#ffffff", font=("Arial", 10, "bold")
        )
        self.ping_label.place(x=90, y=110)

        self.progress = ttk.Progressbar(
            root, orient="horizontal", length=210, mode="indeterminate"
        )
        self.progress.place(x=85, y=140)
        self.progress.place_forget()

        self.button = tk.Button(
            root,
            text="Check Speed ▶",
            width=30,
            bd=0,
            bg="#404042",
            fg="#ffffff",
            pady=5,
            command=self.start_speed_test
        )
        self.button.place(x=85, y=170)

    def start_speed_test(self):
        self.button.config(state=tk.DISABLED)
        self.progress.place(x=85, y=140)
        self.progress.start(10)

        thread = threading.Thread(target=self.check_speed, daemon=True)
        thread.start()

    def check_speed(self):
        try:
            speed_test = speedtest.Speedtest()

            download = speed_test.download()
            upload = speed_test.upload()

            download_speed = round(download / 1_000_000, 2)
            upload_speed = round(upload / 1_000_000, 2)

            speed_test.get_servers([])
            ping = round(speed_test.results.ping, 2)

            self.root.after(
                0, self.update_result,
                download_speed, upload_speed, ping
            )

        except Exception as error:
            self.root.after(0, self.show_error, error)

    def update_result(self, download_speed, upload_speed, ping):
        self.progress.stop()
        self.progress.place_forget()

        self.down_label.config(
            text=f"⏬ Download Speed - {download_speed} Mbps"
        )
        self.up_label.config(
            text=f"⏫ Upload Speed - {upload_speed} Mbps"
        )
        self.ping_label.config(
            text=f"Your Ping - {ping} ms"
        )

        self.button.config(state=tk.NORMAL)

    def show_error(self, error):
        self.progress.stop()
        self.progress.place_forget()
        self.button.config(state=tk.NORMAL)

        messagebox.showerror(
            "Speed Test Error",
            f"Unable to complete speed test.\n\n{error}"
        )


root = tk.Tk()
app = InternetSpeedTestGUI(root)
root.mainloop()
```

## 4. Find IP Address of a Website

```python
import socket

def get_hostname_IP():
    hostname = input(
        "Please enter website address (example: google.com): "
    ).strip()

    try:
        print(f"Hostname: {hostname}")
        print(f"IP Address: {socket.gethostbyname(hostname)}")
    except socket.gaierror as error:
        print(f"Invalid Hostname. Error: {error}")


get_hostname_IP()
```

## Important Notes

- Run the speed test while connected to the internet.
- The GUI uses a background thread so the window remains responsive.
- Do not use `root.iconbitmap("speed.ico")` unless the icon file exists.
- The GUI gets ping from the same `speed_test` object used for download and upload.
